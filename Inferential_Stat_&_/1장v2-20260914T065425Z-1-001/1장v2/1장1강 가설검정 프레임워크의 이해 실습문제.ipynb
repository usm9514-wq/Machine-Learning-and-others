#### 제출 결과

- 가설 설정
- 표본 수와 표본평균

```python
# 필수 1 코드를 작성하세요.

sale_price = df['SalePrice'].dropna()
mu0 = 180000
alpha = 0.05

result = stats.ttest_1samp(
    sale_price,
    popmean=mu0,
    alternative ="two-sided"
)
print(f'표본 수 : {len(sale_price)}')
print(f'sample mean : {sale_price.mean(): ,.2f} dollars')
print(f'stat: {result.statistic: .4f}')
print(f'p-value: {result.pvalue: .4f}')

if result.pvalue <= alpha:
    print('judgement: 귀무가설을 기각합니다.')
else:
    print('judgement: 귀무가설을 기각하지 못합니다')
```

## 필수 2. 양측검정과 단측검정 비교

### 문제 2-1. 평균 품질 점수는 6점과 다른가, 그리고 6점보다 높은가?

#### 문제 설명

주택의 전반적인 품질을 나타내는 `OverallQual`의 기준 점수를 **6점**으로 설정했습니다.  
같은 표본으로 다음 두 연구 질문을 각각 검정하고 결과를 비교하세요.

- 질문 A: 평균 품질 점수가 6점과 **다른가?**
- 질문 B: 평균 품질 점수가 6점보다 **높은가?**

#### 요구사항

1. `OverallQual`에서 결측치를 제거하여 `overall_qual`에 저장하세요.
2. 표본 수와 표본평균을 출력하세요.
3. 질문 A의 가설을 작성하고 `alternative="two-sided"`로 양측검정을 수행하세요.
4. 질문 B의 가설을 작성하고 `alternative="greater"`로 단측검정을 수행하세요.
5. 두 검정의 t통계량과 p-value를 출력하세요.
6. 유의수준 0.05를 기준으로 각 검정의 귀무가설 기각 여부를 출력하세요.
7. 같은 데이터인데 두 p-value가 다르게 계산되는 이유를 설명하세요.

#### 해석 질문

**Q1.** 질문 A와 질문 B의 대립가설은 어떻게 다른가요?  
**Q2.** 양측검정과 단측검정의 결론은 각각 무엇인가요?  
**Q3.** 분석 결과를 먼저 확인한 뒤 p-value를 작게 만들기 위해 단측검정으로 바꿔도 되나요?  
**Q4.** 실제 평균 품질 점수가 6점인데도 검정 결과가 유의하다고 판단했다면 어떤 오류인가요?

#### 제출 결과

- 두 검정의 가설
- 표본 수와 표본평균
- 양측검정 및 단측검정 코드
- 두 검정의 t통계량과 p-value
- 검정별 귀무가설 판단
- Q1~Q4 답변

```python
# 필수 2 코드를 작성하세요.



MU0 = 6       # 귀무가설 기준값
ALPHA = 0.05  # 유의수준
 
 
# ---------------------------------------------------------------
# 1. OverallQual 결측치 제거
# ---------------------------------------------------------------
overall_qual = df["OverallQual"].dropna()
 
 
# ---------------------------------------------------------------
# 2. 표본 수와 표본평균
# ---------------------------------------------------------------
n = len(overall_qual)
sample_mean = overall_qual.mean()
 
print("=" * 55)
print("[기초 통계]")
print(f"표본 수(n)  : {n}")
print(f"표본평균     : {sample_mean:.4f}")
print(f"표본표준편차 : {overall_qual.std(ddof=1):.4f}")
 
 
# ---------------------------------------------------------------
# 3. 질문 A - 양측검정
#    H0: mu = 6   (평균 품질 점수는 6점이다)
#    H1: mu != 6  (평균 품질 점수는 6점이 아니다 - 방향 지정 없음)
# ---------------------------------------------------------------
t_two, p_two = stats.ttest_1samp(overall_qual, popmean=MU0, alternative="two-sided")
 
 
# ---------------------------------------------------------------
# 4. 질문 B - 단측검정
#    H0: mu <= 6  (평균 품질 점수는 6점 이하다)
#    H1: mu >  6  (평균 품질 점수는 6점보다 높다 - 방향 지정)
# ---------------------------------------------------------------
t_one, p_one = stats.ttest_1samp(overall_qual, popmean=MU0, alternative="greater")
 
 
# ---------------------------------------------------------------
# 5. t통계량과 p-value 출력
# ---------------------------------------------------------------
print("=" * 55)
print("[검정 결과]")
print(f"질문 A (양측, mu != {MU0})  t = {t_two:.4f},  p = {p_two:.6f}")
print(f"질문 B (단측, mu >  {MU0})  t = {t_one:.4f},  p = {p_one:.6f}")
 
 
# ---------------------------------------------------------------
# 6. 유의수준 0.05 기준 귀무가설 기각 여부
# ---------------------------------------------------------------
print("=" * 55)
print(f"[판정] 유의수준 alpha = {ALPHA}")
 
for label, p in [("질문 A (양측검정)", p_two), ("질문 B (단측검정)", p_one)]:
    if p < ALPHA:
        print(f"{label}: p = {p:.6f} < {ALPHA} -> 귀무가설 기각")
    else:
        print(f"{label}: p = {p:.6f} >= {ALPHA} -> 귀무가설 기각 실패")
 
 
# ---------------------------------------------------------------
# 7. 같은 데이터인데 두 p-value가 다른 이유
# ---------------------------------------------------------------
print("=" * 55)
print("[7. 같은 데이터인데 p-value가 다른 이유]")
 
if t_two > 0:
    relation = (f"t > 0 이므로 두 꼬리가 대칭이라 정확히 단측 p = 양측 p / 2 가 된다.\n"
                f"실제로 {p_two:.6f} / 2 = {p_two / 2:.6f} 이고, 단측 p = {p_one:.6f} 로 일치한다.")
else:
    relation = (f"t < 0 이라 방향이 대립가설과 반대다. 이 경우 단측 p = 1 - 양측 p / 2 가 되어\n"
                f"오히려 양측보다 커진다 (단측 p = {p_one:.6f}).")
 
print(f"""
t통계량은 두 검정에서 완전히 동일하다 (t = {t_two:.4f}).
데이터가 같으니 당연하다. 달라지는 것은 t분포에서 '어느 영역의 넓이를
p-value로 세느냐'뿐이다.
 
- 양측검정: 대립가설이 방향을 지정하지 않으므로, 관측된 t만큼 극단적인
  경우를 오른쪽 꼬리와 왼쪽 꼬리 양쪽에서 센다.
    p = P(T >= |t|) + P(T <= -|t|)
 
- 단측검정: 대립가설이 '크다'는 방향을 미리 지정했으므로, 오른쪽 꼬리만 센다.
    p = P(T >= t)
 
{relation}
 
즉 p-value 차이는 데이터가 달라서가 아니라, 대립가설을 어떻게 세웠는지에
따라 '증거로 인정할 방향'이 달라졌기 때문이다.
""")
print("=" * 55)
 
```

`GrLivArea`를 사용하여 이 주장을 단일표본 t검정으로 확인하세요.

> 이 문제는 필수 문제에서 연습한 가설 설정과 단측검정 절차를 새로운 변수에 적용하는 문제입니다.

#### 요구사항

1. `GrLivArea`에서 결측치를 제거하여 분석 표본을 준비하세요.

#### 해석 질문

**Q1.** 이 문제에서 적절한 `alternative` 값은 무엇인가요?  
**Q2.** p-value는 귀무가설이 참일 확률을 의미하나요?  
**Q3.** 실제 평균이 1,550보다 작은데 귀무가설을 기각하지 못했다면 어떤 오류인가요?

#### 제출 결과

- 가설 설정과 검정 방향 선택 이유
- 표본 수와 표본평균
- 검정 코드
- t통계량과 p-value
- 귀무가설 판단과 분석 결론
- Q1~Q3 답변

```python
# 과제 1 코드를 작성하세요.

from scipy import stats

# 1) 결측치 제거
x = df['GrLivArea'].dropna()

# 2) 표본 수와 표본평균
n, xbar = len(x), x.mean()
print(f"표본 수 n = {n}")
print(f"표본평균 = {xbar:.4f}")

# 5) 단일표본 t검정 (좌측 단측)
t_stat, p_value = stats.ttest_1samp(x, popmean=1550, alternative='less')

# 6) 소수점 넷째 자리까지 출력
print(f"t-통계량 = {t_stat:.4f}")
print(f"p-value  = {p_value:.4f}   (지수표기: {p_value:.4e})")

# 7) 유의수준 0.05 기준 판단
alpha = 0.05
if p_value < alpha:
    print(f"p-value < {alpha} → 귀무가설 기각")
else:
    print(f"p-value >= {alpha} → 귀무가설 기각 실패")
```

**Q3.** 실제 평균이 1,550보다 작은데 귀무가설을 기각하지 못했다면 어떤 오류인가요?
Type II error로 봐야한다. 

왜냐하면 
μ < 1550 (H₀ 거짓) 이부분에서 못 쳐내서 잘못(실제로 마지막에 단계에 부분 확인)
→ 그런데 표본에서 계산한 p-value가 0.05 이상으로 나옴
→ 규칙대로 "기각 못 함" 판정

2. 귀무가설의 기각 여부는 무엇을 기준으로 판단했나요?
-> P-value와 사전에 설정한 유의수준(0.05)를 비교
-> p-value <= 0.05이면 귀무가설을 기각하고, p-value > 0.05면 기각하지 못함

3. 양측검정과 단측검정은 연구 질문에서 어떤 차이가 있나요?
-> 양측검정은 기준과 다르지 확인하고
-> 단측검정은 기준보다 크거나 작은 특정 방향을 확인한다.

4. 귀무가설을 기각하지 못한다는 것은 무엇을 의미하나요?
귀무가설을 참이라고 증명한 것이 아니라 대립가설을 지지할 증거 충분하지 않다는 의미입니다.

```python
## user_ids = [101,203,305,407,509] / target = 305

def solution(user_ids, target):
    left = 0                    # left = 0 / right 5-1=4
    right = len(user_ids) - 1   
    while left <= right:  #0 <=4 ->참  = 실행
        mid = (left + right) // 2   # mid = (0+4)//2 = 2
        if user_ids[mid] == target:
            return mid
        elif user_ids[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1

print(solution([101,203,305,407,509],305))  # Output: 2
print(solution([10,20,30,40,50,60],60))  # Output: 5
print(solution([2,4,6,8,10],7))  # Output: -1
```

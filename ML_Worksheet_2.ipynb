```python
import numpy as np

x = np.array([1, 2, 3, 4, 5])
y = np.array([50, 60, 70, 80, 90])

X = np.column_stack([
    x,
    np.ones(len(x))
])

W = np.linalg.inv(X.T @ X) @ X.T @ y

weight = W[0]
bias = W[1]

print(f"Weight: {weight:.1f}")
print(f"Bias: {bias:.1f}")
```

```python
import numpy as np

x = np.array([1.0, 2.0, 3.0, 4.0])
y = np.array([20.0, 30.0, 40.0, 50.0])

weight = 0.0
bias = 0.0

learning_rate = 0.05
epochs = 1000

n = len(x)

for epoch in range(epochs):
    y_pred = weight * x + bias

    error = y_pred - y
    loss = np.mean(error ** 2)

    weight_gradient = (2 / n) * np.sum(error * x)
    bias_gradient = (2 / n) * np.sum(error)

    weight = weight - learning_rate * weight_gradient
    bias = bias - learning_rate * bias_gradient

print(f"Weight: {weight:.2f}")
print(f"Bias: {bias:.2f}")
print(f"Loss: {loss:.4f}")
```

```python
import numpy as np

from sklearn.linear_model import Lasso

X = np.array([
    [1, 5, 2, 10],
    [2, 5, 1, 20],
    [3, 5, 3, 30],
    [4, 5, 2, 40],
    [5, 5, 1, 50],
    [6, 5, 3, 60]
])

y = np.array([12, 22, 31, 42, 51, 62])

lasso = Lasso(alpha=1.0)
lasso.fit(X, y)

print("Lasso 가중치:", np.round(lasso.coef_, 2))

for i, weight in enumerate(lasso.coef_):
    if weight == 0:
        print(f"{i + 1}번째 Feature의 가중치는 0입니다.")
```

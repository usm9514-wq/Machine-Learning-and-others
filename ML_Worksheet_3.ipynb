```python
class RequestCounter:
    def __init__(self):
        self.counts = {}  # 사용자별 요청 횟수를 저장할 딕셔너리


    def add_request(self, user):
        if user in self.counts:
            self.counts[user] += 1
        else:
            self.counts[user] = 1

    def get_count(self, user):
        return self.counts[user] 


def solution(requests):
    counter = RequestCounter()  

    for user in requests:
        counter.add_request(user)
    
    best_user = requests[0]
    best_count = counter.get_count(best_user)

    for user in requests:
        current_count = counter.get_count(user)
        if current_count > best_count:
            best_user = user
            best_count = current_count

    return best_user


print(solution(["alice", "bob", "alice", "charlie", "alice", "bob"]))
print(solution(["tom", "jane", "jane", "tom", "tom"]))
print(solution(["mina", "jisu", "mina", "jisu"]))
```

```python
import numpy as np
from sklearn.linear_model import LogisticRegression

X = np.array([
    [1],
    [2],
    [3],
    [5],
    [7],
    [9]
])

y = np.array([0, 0, 0, 1, 1, 1])

model = LogisticRegression()
model.fit(X, y)

new_customer = np.array([[6]])

probability = model.predict_proba(new_customer)
prediction = model.predict(new_customer)

print("구매 확률:", round(probability[0][1], 3))
print("예측 클래스:", prediction[0])
```

```python
import numpy as np

logits = np.array([1.0, 2.5, 0.5])

exp_logits = np.exp(logits)
probabilities = exp_logits / np.sum(exp_logits)

prediction = np.argmax(probabilities)

print("각 클래스 확률:", np.round(probabilities, 3))
print("예측 클래스:", prediction)
print("확률의 합:", round(probabilities.sum(), 3))
```

```python
class EditHistory:
    def __init__(self):
        self.history = []
    def add(self, action):
        self.history.append('action')
    def undo(self):
        if self.history:
            self.history.pop()

    def get_history(self):
        return self.history

def solution(commands):
    edit_history = EditHistory()

    for command in commands:
        if command == "UNDO":
            edit_history.undo()
        else:
            edit_history.add(command)
    return edit_history.get_history()


print(solution(["밝기 조절", "배경 제거", "크기 조절", "UNDO"]))
print(solution(["로그인", "검색", "UNDO", "결제"]))
print(solution(["UNDO", "텍스트 추가", "UNDO", "UNDO"]))
```

```python
from sklearn.metrics import accuracy_score, recall_score

y_true = [1] * 20 + [0] * 980
y_pred = [0] * 1000

acc = accuracy_score(y_true, y_pred)
rec = recall_score(y_true, y_pred)
auc = roc_auc_score(y_true, y_pred)

# 3. 결과 출력
print(f"Accuracy : {acc:.4f}")
print(f"Recall   : {rec:.4f}")
print(f"ROC-AUC  : {auc:.4f}")

```

```python
from collections import deque

class CustomerQueue:
    def __init__(self):
        self.queue = deque()
        pass

    def add(self, name):
        self.queue.append(name)
        pass

    def call(self):
        return self.queue.popleft()
        pass

    def count(self):
        return len(self.queue)
        pass


queue1 = CustomerQueue()
queue1.add('민수')
print(queue1.call())
```

이런 분석에는 비지도 학습이 필요 한다 왜냐하면 1개월 방문 횟수로는 매우 짧은 시간이기에 이것을 토대로 기준을 만들고 이걸 기반으로 결과를 내면 편향적인 데이터를 도출 할 수 잇기 때문이다.

```python
import pandas as pd

from sklearn.cluster import KMeans


customers = pd.DataFrame({
    "order_count": [1, 2, 3, 4, 12, 14, 15, 17],
    "total_purchase": [20, 30, 35, 50, 200, 230, 250, 280]
})

model = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

customers["cluster"] = model.fit_predict(
    customers[["order_count", "total_purchase"]]
)

print(customers)
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA


data = pd.DataFrame({
    "study_hours": [1, 2, 3, 5, 7, 9],
    "problem_count": [10, 18, 30, 48, 70, 90],
    "mock_score": [45, 50, 58, 70, 82, 92]
})

scaler = StandardScaler()
X_scaled = scaler.fit_transform(data)

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)


print(data.shape)
print(X_pca.shape)
print(pca.explained_variance_ratio_)
```

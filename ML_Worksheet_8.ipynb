```python
class ShoppingPlanner:
    def __init__(self, budget):
        self.budget = budget
        self.prices = []
        # 현재 객체의 예산을 저장하세요.

        # 상품 가격을 저장할 빈 리스트를 만드세요.
        pass

    def add_product(self, price):
        self.prices.append(price)
        # 현재 객체의 상품 가격 리스트에 price를 추가하세요.
        pass

    def find_best_pair(self):
        # 서로 다른 상품 두 개를 선택하는 모든 경우를 확인하세요.
        best_price = 0

        # 예산을 넘지 않는 경우 중 가장 큰 가격 합을 찾으세요.
        # 첫번째 루프는 0 부터 리스트 길이 -1  
        for i in range(len(self.prices)):
            for j in range(i+1, len(self.prices)):
                ## 모든 가능한 경우의 수에 해당하는 i , j
                total_price = self.prices[i] + self.prices[j]
            
                if total_price <= self.budget:
                    if total_price > best_price:
                        best_price = total_price
        return best_price


planner_a = ShoppingPlanner(10000)

planner_a.add_product(3000)
planner_a.add_product(5000)
planner_a.add_product(7000)
planner_a.add_product(2000) 
## planner_a # budget은 1만, prcies [3000, 5000, 7000, 2000]

# 모든 가능한 경우의 수에 대해 반복
# -1. 두개의 상품의 인덱스 선택
# -2. 선택한 인덱스 기준으로 두개의 상품 가격 확인 상품 가격의 합 구함
# -3. 예산 범위 안에 들어오는지 확인 + 이전에 찾아놓은 상품 



```

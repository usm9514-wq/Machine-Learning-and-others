```python
class CoinPayment:
    def __init__(self, amount):
        self.amount = amount
        self.coins = [500,100,50,10]        

    def min_coins(self):
        total_count = 0
        remaining = self.amount
        for coin in self.coins:
            count = remaining // coin
            total_count += count
            remaining %= coin

        return total_count


payment_a = CoinPayment(1260)
payment_b = CoinPayment(800)
payment_c = CoinPayment(70)

print(payment_a.min_coins())
print(payment_b.min_coins())
print(payment_c.min_coins())
```

```python
from scipy import stats

purchase = [
    52000, 55000, 49000, 57000, 54000,
    51000, 56000, 53000, 50000, 55000
]

t_stat, p_value = stats.ttest_1samp(
    purchase,
    popmean=50000
)

print(f"표본 평균: {sum(purchase) / len(purchase):.0f}원")
print(f"p-value: {p_value:.4f}")
```

```python
from scipy import stats

delivery_days = [
    2.8, 2.5, 2.7, 3.0, 2.6,
    2.9, 2.4, 2.7, 2.8, 2.6
]

# H0: 평균 배송 시간 = 3일
# H1: 평균 배송 시간 ≠ 3일

t_stat, p_value = stats.ttest_1samp(
    delivery_days,
    popmean=3
)

print(f"t = {t_stat:.3f}")
print(f"p = {p_value:.4f}")

if p_value < 0.05:
    print("귀무가설을 기각합니다.")
    print("평균 배송 시간이 기존 3일과 다르다고 볼 통계적 근거가 있습니다.")
else:
    print("귀무가설을 기각할 수 없습니다.")
    print("평균 배송 시간이 기존 3일과 다르다고 볼 통계적 근거가 충분하지 않습니다.")
```

```python
from statsmodels.stats.power import TTestIndPower

analysis = TTestIndPower()

power_30 = analysis.power(
    effect_size=0.4,
    nobs1=30,
    alpha=0.05
)

power_100 = analysis.power(
    effect_size=0.4,
    nobs1=100,
    alpha=0.05
)

print(f"그룹당 30명일 때 검정력: {power_30:.3f}")
print(f"그룹당 100명일 때 검정력: {power_100:.3f}")
```

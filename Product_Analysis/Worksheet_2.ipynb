```python
import pandas as pd


df = pd.read_csv(r'C:\Machine_Learning_worksheet\Product_Analysis\learning_product_event_log.csv')


# 날짜 형식 변환
df["event_time"] = pd.to_datetime(df["event_time"])
df["signup_date"] = pd.to_datetime(df["signup_date"])

# 전체 사용자
total_users = df["user_id"].nunique()

# Acquisition: 서비스 방문
acquisition_users = df.loc[
    df["event_name"] == "visit",
    "user_id"
].nunique()

# Activation: 실제 학습 시작
activation_users = df.loc[
    df["event_name"] == "start_lesson",
    "user_id"
].nunique()

# Retention: 가입 후 정확히 7일째 재방문
visit_df = df[df["event_name"] == "visit"].copy()

visit_df["day"] = (
    visit_df["event_time"].dt.normalize()
    - visit_df["signup_date"].dt.normalize()
).dt.days

retention_users = visit_df.loc[
    visit_df["day"] == 7,
    "user_id"
].nunique()

# Revenue: 유료 구독
revenue_users = df.loc[
    df["event_name"] == "subscribe",
    "user_id"
].nunique()

# Referral: 강의 공유
referral_users = df.loc[
    df["event_name"] == "share",
    "user_id"
].nunique()

# AARRR 결과 정리
aarrr = pd.DataFrame({
    "stage": [
        "Acquisition",
        "Activation",
        "Retention",
        "Revenue",
        "Referral"
    ],
    "users": [
        acquisition_users,
        activation_users,
        retention_users,
        revenue_users,
        referral_users
    ]
})

aarrr["rate"] = (
    aarrr["users"] / total_users * 100
).round(1)

print(aarrr)

```

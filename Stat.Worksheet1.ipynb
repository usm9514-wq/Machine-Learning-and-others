```python
payments = [3, 5, 2, 7]

dp = [0] * len(payments)

dp[0] = payments[0]
dp[1] = max(payments[0], payments[1])

for i in range(2, len(payments)):
    rest = dp[i - 1]
    work = dp[i - 2] + payments[i]

    dp[i] = max(rest, work)

print(dp)
print(dp[-1])
```

```python
class PartTimeWorker:
    def __init__(self, payments):
        self.payments = payments
    
    def max_payment(self):
        dp = [0] * len(self.payments)

        dp[0] = self.payments[0]
        dp[1] = max(self.payments[0], self.payments[1])

        for i in range(2, len(self.payments)):
            rest = dp[i-1]
            work = dp[i-2] + self.payments[i]

            dp[i] = max(rest,work)
        
        return dp[-1]

worker_a = PartTimeWorker([3, 5, 2, 7])
worker_b = PartTimeWorker([4, 1, 6])
worker_c = PartTimeWorker([5, 10, 4, 11, 3])



print(worker_a.max_payment())
print(worker_b.max_payment())
print(worker_c.max_payment())


    
```

```python

from scipy import stats

weights = [505, 498, 503, 507, 501, 504, 499, 506, 502, 505]


t_stat, p_value = stats.ttest_1samp(
    weights,
    popmean=500,
)

print(f"t 통계량: {t_stat:.3f}")
print(f"p-value: {p_value:.3f}")
```

```python
company_a = [28, 31, 29, 30, 27, 32, 29, 28, 30, 31]
company_b = [35, 33, 36, 34, 32, 37, 35, 34, 36, 33]

t_stat, p_value = stats.ttest_ind(
    company_a,
    company_b,
    equal_var=False
)

print(f"t 통계량: {t_stat:.3f}")
print(f"p-value: {p_value:.3f}")
```

```python
class NetworkExplorer:
    def __init__(self, network):
        self.network = network

    def count_connected(self, start):
        stack = [start]
        visited = set()

        while stack:
            current = stack.pop()

            if current not in visited:
                visited.add(current)

                for next_node in self.network[current]:
                    if next_node not in visited:
                        stack.append(next_node)

        return len(visited)

network_a = {
    1: [2, 3],
    2: [1, 4],
    3: [1],
    4: [2],
    5: []
}

network_b = {
    1: [2],
    2: [1, 3],
    3: [2]
}

network_c = {
    1: [],
    2: [3],
    3: [2]
}

explorer_a = NetworkExplorer(network_a)
explorer_b = NetworkExplorer(network_b)
explorer_c = NetworkExplorer(network_c)

print(explorer_a.count_connected(1))
print(explorer_b.count_connected(2))
print(explorer_c.count_connected(1))
```

```python
from scipy import stats

method_a = [68, 72, 70, 74, 71]
method_b = [75, 78, 76, 80, 77]
method_c = [82, 85, 83, 86, 84]


f_stat , p_value = stats.f_oneway(
    method_a,
    method_b,
    method_c
)


print(f"F 통계량: {f_stat:.3f}")
print(f"p-value: {p_value:.3f}")


```

```python
from scipy import stats

table = [
    [50, 30],
    [25, 45],
    [20, 30]
]

chi2, p_value, dof, expected = stats.chi2_contingency(table)

print(f"카이제곱 통계량: {chi2:.3f}")
print(f"p-value: {p_value:.3f}")
```

```python
import numpy as np

study_hours = np.array([1, 2, 3, 4, 5, 6, 7, 8])
scores = np.array([52, 58, 63, 69, 74, 81, 86, 92])

from sklearn.linear_model import LinearRegression

X = study_hours.reshape(-1, 1)
y = scores

model = LinearRegression()
model.fit(X, y)

print("절편:", model.intercept_)
print("회귀계수:", model.coef_[0])
```

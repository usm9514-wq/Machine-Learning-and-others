대조군이 40대 이상인데 실험군이 갑자기 20~30대 사용자로 타겟이 controlled되어 있지 않다.
40 대 이상 사용자로 통일 해야ㅎ한다

대조군: 기존 결제 화면
할인 쿠폰 없음

실험군: 새로운 결제 화면
20% 할인 쿠폰 제공

현재 대조군의 조건과 실험군 조건에 두가지 변수가 동시에 넣었기 때문에 오류이며 제대로 하려면 기존 결제 화면에서 할일 쿠폰을 쓴거를 확인해야한다.


```python
import math

from statsmodels.stats.power import NormalIndPower
from statsmodels.stats.proportion import proportion_effectsize

baseline = 0.10
mde = 0.01
target = baseline + mde

analysis = NormalIndPower()

sample_size = analysis.solve_power(
    effect_size=abs(proportion_effectsize(baseline, target)),
    alpha=0.05,
    power=0.80,
    ratio=1.0,
    alternative="two-sided"
)

sample_size = math.ceil(sample_size)

print(sample_size)
```
